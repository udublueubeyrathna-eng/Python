6th Question

Select the Numerical Variables

```python
numerical_columns = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age"
]
```

Detect Outliers Using the IQR Method

```python
print("Outlier Detection Using IQR\n")

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - (1.5 * IQR)
    upper_limit = Q3 + (1.5 * IQR)

    outliers = (
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ).sum()

    print(f"{column}")
    print(f"Lower Limit : {lower_limit:.2f}")
    print(f"Upper Limit : {upper_limit:.2f}")
    print(f"Outliers    : {outliers}\n")
```

Visualize the Outliers Before Treatment

```python
plt.figure(figsize=(14,8))

df[numerical_columns].boxplot()

plt.title("Boxplots Before Outlier Treatment")

plt.xticks(rotation=45)

plt.show()
```

Treat the Outliers Using the IQR Method

```python
# Create a copy of the cleaned dataset
df_treated = df.copy()

print("Treating Outliers Using IQR Capping\n")

for column in numerical_columns:

    Q1 = df_treated[column].quantile(0.25)
    Q3 = df_treated[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - (1.5 * IQR)
    upper_limit = Q3 + (1.5 * IQR)

    df_treated[column] = df_treated[column].clip(
        lower_limit,
        upper_limit
    )

# Recalculate the engineered feature after treating
# its source variables
df_treated["DiabetesRiskScore"] = (
    (0.5 * df_treated["Glucose"]) +
    (0.3 * df_treated["BMI"]) +
    (0.2 * df_treated["Age"])
)

print("Outlier treatment completed successfully.")
```

Check Whether the Treatment Worked

```python
# Include the engineered feature for final verification
all_numerical_columns = numerical_columns + ["DiabetesRiskScore"]

print("Remaining Outliers After Treatment\n")

for column in all_numerical_columns:

    Q1 = df_treated[column].quantile(0.25)
    Q3 = df_treated[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - (1.5 * IQR)
    upper_limit = Q3 + (1.5 * IQR)

    outliers = (
        (df_treated[column] < lower_limit) |
        (df_treated[column] > upper_limit)
    ).sum()

    print(f"{column}: {outliers}")
```

Visualize the Dataset After Treatment

```python
plt.figure(figsize=(14,8))

df_treated[all_numerical_columns].boxplot()

plt.title("Boxplots After Outlier Treatment")

plt.xticks(rotation=45)

plt.show()
```

Compare the Dataset Before and After Treatment

```python
print("Summary Statistics After Outlier Treatment")

print(df_treated[all_numerical_columns].describe())
```

```python
# Create the final dataset for Tableau
tableau_df = df_treated.copy()

# Add a unique identifier for each patient record
tableau_df.insert(
    0,
    "PatientID",
    range(1, len(tableau_df) + 1)
)

print("First Five Rows of the Tableau Dataset")
print(tableau_df.head())
```
